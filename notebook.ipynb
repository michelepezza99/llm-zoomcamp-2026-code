```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from openai import OpenAI
openai_client = OpenAI()
```

```python
def llm(prompt):
    response = openai_client.responses.create(
        model="gpt-4o-mini",
        input=prompt
    )
    return response.output_text
```

```python
question = "I just discovered the course. Can I join now?"
answer = llm(question)
print(answer)
```
